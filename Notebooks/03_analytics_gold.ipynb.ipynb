**Use Silver Delta table to create DataFrame**
```
1. Load Silver delta table to create DataFrame.
2. Apply GroupBy, aggregation logics to gather metrices.
3. Use cleaned DataFrame to create Gold Delta Table.
```

````python
silver_df=spark.table("silver_taxi_trips")
display(silver_df)
````

````python
from pyspark.sql.functions import *

# gold_df=silver_df.withColumn("pickup_date",to_date(col("tpep_pickup_datetime"))).groupBy(to_date(col("tpep_pickup_datetime"))).agg(sum("trip_distance").alias("total_distance"),sum("fare_amount").alias("total_fare"),sum("tip_amount").alias("total_tip")).orderBy("pickup_date")

gold_df=silver_df.groupBy(to_date(col("tpep_pickup_datetime")).alias("pickup_date")).agg(sum("trip_distance").alias("total_distance"),sum("fare_amount").alias("total_fare"),sum("tip_amount").alias("total_tip")).orderBy("pickup_date")

display(gold_df)
````

**Rename column to create Gold delta table**
```
1. Rename column to "pickup_date" to create delta table from aggregated DataFrame.
```

````python
# gold_df.withColumnRenamed("to_date(tpep_pickup_datetime)", "pickup_date").write.format("delta").mode("overwrite").saveAsTable("gold_taxi_trips")

gold_df.write.format("delta").mode("overwrite").saveAsTable("gold_taxi_trips")
````
