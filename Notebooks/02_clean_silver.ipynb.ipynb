**Bronze Delta table to Silver Delta table**
```
🥉 Bronze → 🥈 Silver
(Delta)      (Delta) 
```

**Use Bronze Delta table to create DataFrame**
```
1. Load Bronze delta table to create DataFrame.
2. Apply filtering to clean DataFrame.
3. Use cleaned DataFrame to create Silver Delta Table.
```

````python
from pyspark.sql.functions import col

bronze_df = spark.table("bronze_taxi_trips")

# display(bronze_df)

clean_df = bronze_df.filter(col("trip_distance") > 0).filter(col("total_amount") > 0)

# display(clean_df)

clean_df.write.format("delta").mode("overwrite").saveAsTable("silver_taxi_trips")

# clean_df.head(5)
````
