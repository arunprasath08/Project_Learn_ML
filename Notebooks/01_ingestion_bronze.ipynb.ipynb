# 🏗️  Data Lakehouse for NYC Taxi trip with Yellow line trip data

**Building analytics and deriving insights using production-grade data architecture using Taxi trip data**

## 🏛️ Lakehouse Architecture Following Databricks Reference

This project implements the **Medallion Architecture** with **Unity Catalog integration**:

```
Data Sources → Object Storage → 🥉 Bronze → 🥈 Silver → 🥇 Gold → Consumption Layer
(NYC Taxi trips)   (Workspace)    (Delta)    (Delta)    (Delta)  (Analytics/ML)
```

**Parquet data file to DataFrame to Bronze Delta table**
```
Data Sources → Object Storage → 🥉 Bronze 
(NYC Taxi trips)   (Workspace)    (Delta) 
```

**Load Parquet data file into DataFrame and create Bronze Delta table**


````python
df = spark.read.option("header", "true").parquet(
    "/Volumes/workspace/nyc_taxi/yellow_trips_01_2025/yellow_tripdata_2025-01.parquet"
)
# df.display()
# df.head(10)
df.write.format("delta").mode("overwrite").saveAsTable("bronze_taxi_trips")
````
